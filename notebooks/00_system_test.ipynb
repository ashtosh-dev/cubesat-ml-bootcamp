# 00 System and Environment Check 

Run these cells before the bootcamp. This notebook checks the active Python environment, important packages, TensorFlow execution, and CIFAR-10 availability.


```python
import sys
import importlib.metadata as metadata

print("Python executable:", sys.executable)
print("Python version:", sys.version.split()[0])

packages = [
    "tensorflow", "numpy", "matplotlib", "scikit-learn",
    "jupyterlab", "ipykernel", "protobuf"
]
for package in packages:
    try:
        print(f"{package}: {metadata.version(package)}")
    except metadata.PackageNotFoundError:
        print(f"{package}: NOT INSTALLED")
```

```python
import tensorflow as tf

print("TensorFlow:", tf.__version__)
gpus = tf.config.list_physical_devices("GPU")
print("Detected GPU devices:", gpus)

# A small operation verifies that TensorFlow can execute.
with tf.device("/GPU:0" if gpus else "/CPU:0"):
    a = tf.random.normal((512, 512))
    b = tf.random.normal((512, 512))
    result = tf.matmul(a, b)

print("Matrix multiplication shape:", result.shape)
print("Execution device:", result.device)
print("TensorFlow computation: PASS")
```
